```python
from pathlib import Path
import cv2

all_widths = []
all_heights = []

image_files = list(TRAIN_IMAGES.glob("*.png"))

for image_path in image_files:

    image = cv2.imread(str(image_path))

    if image is None:
        continue

    img_h, img_w = image.shape[:2]

    label_path = TRAIN_LABELS / (image_path.stem + ".txt")

    if not label_path.exists():
        continue

    with open(label_path, "r") as f:
        labels = f.readlines()

    for line in labels:

        cls, xc, yc, w, h = map(float, line.split())

        all_widths.append(w * img_w)
        all_heights.append(h * img_h)

print("Total Objects :", len(all_widths))
```

```python
import numpy as np

print("Average Width :", np.mean(all_widths))
print("Average Height:", np.mean(all_heights))

print("Minimum Width :", np.min(all_widths))
print("Maximum Width :", np.max(all_widths))

print("Minimum Height:", np.min(all_heights))
print("Maximum Height:", np.max(all_heights))
```

```python
sizes = []

for w, h in zip(all_widths, all_heights):
    sizes.append(w * h)

sizes = np.array(sizes)

print("Average Area :", np.mean(sizes))
print("Median Area  :", np.median(sizes))
```

```python
from PIL import Image
import numpy as np
import cv2

def load_image(image_path):
    """
    Load an image using Pillow and convert it to OpenCV (BGR) format.
    Works with JPG, JPEG, PNG, WebP, etc.
    """
    pil_img = Image.open(image_path).convert("RGB")
    image = cv2.cvtColor(np.array(pil_img), cv2.COLOR_RGB2BGR)
    return image
```

```python
from pathlib import Path

TEST_IMAGES = PROJECT_PATH / "test_images"

image_files = sorted(TEST_IMAGES.glob("*"))

for image_path in image_files:

    image = load_image(image_path)

    h, w = image.shape[:2]

    print(f"{image_path.name}")
    print(f"Width : {w}")
    print(f"Height: {h}")
    print("-"*40)
```

```python
from ultralytics import YOLO

MODEL = YOLO(PROJECT_PATH / "dataset" / "models" / "best_640.pt")
```

```python
import numpy as np

for image_path in image_files:

    print("=" * 60)
    print(image_path.name)

    # Load using Pillow
    image = load_image(image_path)

    # Predict using the image array
    results = MODEL.predict(
        source=image,
        imgsz=640,
        conf=0.25,
        verbose=False
    )

    if len(results) == 0:
        print("No prediction returned")
        continue

    result = results[0]

    if len(result.boxes) == 0:
        print("No detections")
        continue

    widths = []
    heights = []

    for box in result.boxes:
        x1, y1, x2, y2 = box.xyxy[0].tolist()
        widths.append(x2 - x1)
        heights.append(y2 - y1)

    print("Objects Detected :", len(widths))
    print("Average Width    :", round(np.mean(widths), 2))
    print("Average Height   :", round(np.mean(heights), 2))
```
