```python
from pathlib import Path
import yaml
import cv2
import matplotlib.pyplot as plt
import ultralytics
from ultralytics import YOLO
```

```python
print(ultralytics.__version__)
```

```python
PROJECT_ROOT = Path.cwd().parent
DATA_YAML = PROJECT_ROOT / "dataset" / "data.yaml"
with open(DATA_YAML, "r") as file:
    data = yaml.safe_load(file)
print(data)
```

```python
print(f"Training Images   : {data['train']}")
print(f"Validation Images : {data['val']}")
print(f"Test Images       : {data['test']}")
print(f"\nNumber of Classes : {data['nc']}")
print("\nClass Names")
for class_id, class_name in data["names"].items():
    print(f"{class_id:2} : {class_name}")
```

```python
# LOAD PRETRAINED YOLO MODEL

model = YOLO("yolov8n.pt")
print("YOLOv8 Nano Model Loaded Successfully")
```

```python
# START TRAINING

print("Model              : YOLOv8 Nano")
print("Epochs             : 5")
print("Image Size         : 512")
print("Batch Size         : 2")
print("Workers            : 0")
print("Device             : CPU")
print("Cache              : False")
```

```python
# TRAIN MODEL

results = model.train(
    data=str(DATA_YAML),
    epochs=1,
    imgsz=320,
    batch=1,
    workers=0,
    device="cpu",
    cache=False,
    name="warehouse_inventory"
)
```

```python
# TRAINING COMPLETED

print("Model training finished successfully.")
print("\nThe trained model and results are saved in:")
print("runs/detect/warehouse_inventory/")
```
